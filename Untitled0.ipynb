Dry Bean Variety Classification


Performance comparison of Logistic Regression, KNN, SVM and Decision Tree

Aim: Classify dry bean varieties using numerical measurements and compare four algorithms.

Dataset: UCI Dry Bean Dataset, with 16 numerical features and seven bean classes.

Procedure: Dataset collection → inspection → preprocessing → train/test split → model building → evaluation and visualization

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from ucimlrepo import fetch_ucirepo
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, classification_report, confusion_matrix, ConfusionMatrixDisplay
RANDOM_STATE = 42
sns.set_theme()
```

```python
dry_bean = fetch_ucirepo(id=602)
X = dry_bean.data.features.copy()
y = dry_bean.data.targets.iloc[:, 0].copy()
df = X.copy()
df["Class"] = y
print("Dataset shape:", df.shape)
display(df.head())
df.info()
print("\\nMissing values per column:")
print(df.isnull().sum())
print("\\nClass counts:")
print(y.value_counts())
```

```python
print("Duplicate rows:", df.duplicated().sum())
df = df.drop_duplicates().reset_index(drop=True)
X = df.drop(columns=["Class"])
y = df["Class"]

plt.figure(figsize=(9, 4))
sns.countplot(data=df, x="Class", order=df["Class"].value_counts().index)
plt.title("Distribution of Dry Bean Classes")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()

plt.figure(figsize=(12, 9))
sns.heatmap(X.corr(), cmap="coolwarm", center=0)
plt.title("Correlation Between Dry Bean Features")
plt.tight_layout()
plt.show()
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=RANDOM_STATE, stratify=y
)
print("Training feature shape:", X_train.shape)
print("Testing feature shape:", X_test.shape)
```

```python
models = {
    "Logistic Regression": Pipeline([("scaler", StandardScaler()),
        ("model", LogisticRegression(max_iter=2000, random_state=RANDOM_STATE))]),
    "KNN": Pipeline([("scaler", StandardScaler()),
        ("model", KNeighborsClassifier(n_neighbors=5))]),
    "SVM": Pipeline([("scaler", StandardScaler()),
        ("model", SVC(kernel="rbf", C=3.0))]),
    "Decision Tree": DecisionTreeClassifier(random_state=RANDOM_STATE)
}
results, predictions = [], {}
for name, model in models.items():
    print("Training:", name)
    model.fit(X_train, y_train)
    pred = model.predict(X_test)
    predictions[name] = pred
    results.append({
        "Algorithm": name,
        "Accuracy": accuracy_score(y_test, pred),
        "Precision (macro)": precision_score(y_test, pred, average="macro", zero_division=0),
        "Recall (macro)": recall_score(y_test, pred, average="macro", zero_division=0),
        "F1-score (macro)": f1_score(y_test, pred, average="macro", zero_division=0)
    })
results_df = pd.DataFrame(results).sort_values("F1-score (macro)", ascending=False)
display(results_df.round(4))
```

```python
results_df.set_index("Algorithm")[["Accuracy", "F1-score (macro)"]].plot(
    kind="bar", figsize=(10, 5), ylim=(0, 1), rot=20)
plt.title("Performance Comparison of Classification Algorithms")
plt.ylabel("Score")
plt.tight_layout()
plt.show()

best_name = results_df.iloc[0]["Algorithm"]
best_pred = predictions[best_name]
print("Best model by macro F1-score:", best_name)
print(classification_report(y_test, best_pred, zero_division=0))
labels = sorted(y.unique())
cm = confusion_matrix(y_test, best_pred, labels=labels)
ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=labels).plot(
    xticks_rotation=45, colorbar=False)
plt.title(f"Confusion Matrix — {best_name}")
plt.tight_layout()
plt.show()
```
